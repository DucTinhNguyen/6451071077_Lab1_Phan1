```python

print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

# Linear algebra
import numpy as np

# Data processing
import pandas as pd

# Data visualization
import seaborn as sns
%matplotlib inline
from matplotlib import pyplot as plt
from matplotlib import style

# Đọc dữ liệu từ file bank-full(1).csv
df = pd.read_csv(
    "bank-full.csv",
    sep=";",
    dtype={
        "age": np.int64,
        "balance": np.int64,
        "day": np.int64,
        "duration": np.int64,
        "campaign": np.int64,
        "pdays": np.int64,
        "previous": np.int64
    }
)

```

```python

print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

FacetGrid = sns.FacetGrid(df, row='loan', height=4.5, aspect=1.6)
FacetGrid.map(sns.pointplot, 'education', 'duration', 'y',
              palette=['blue', 'gray'],
              hue_order=['yes', 'no'])
FacetGrid.add_legend()
```

```python

print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

df['subscribed'] = df['y'].map({'yes': 1, 'no': 0})

sns.barplot(x='education', y='subscribed', data=df)

plt.xlabel('Education')
plt.ylabel('Subscription rate')
plt.title('Education vs Subscription Rate')
plt.show()
```

```python

print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

grid = sns.FacetGrid(df, col='y', row='loan', height=4.2, aspect=1.6)
grid.map(plt.hist, 'age', alpha=.5, bins=20)
grid.add_legend()
plt.show()
```
