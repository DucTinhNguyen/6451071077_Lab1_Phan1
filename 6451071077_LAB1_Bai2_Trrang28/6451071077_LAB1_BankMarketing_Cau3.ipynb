```python

print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

# Linear algebra
import numpy as np

# Data processing
import pandas as pd

# Data visualization
import seaborn as sns
%matplotlib inline
from matplotlib import pyplot as plt
from matplotlib import style

# Đọc dữ liệu từ file bank-full(1).csv
df = pd.read_csv(
    "bank-full.csv",
    sep=";",
    dtype={
        "age": np.int64,
        "balance": np.int64,
        "day": np.int64,
        "duration": np.int64,
        "campaign": np.int64,
        "pdays": np.int64,
        "previous": np.int64
    }
)

```

```python
print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

# Đọc dữ liệu nếu chưa có biến df
# Bank Marketing thường dùng dấu ; để phân cách các cột
# df = pd.read_csv("bank.csv", sep=";")

# Chuẩn hóa tên cột thành chữ thường
df.columns = df.columns.str.lower()

# Tạo hai biểu đồ nằm cạnh nhau
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(15, 5))

# Biểu đồ 1: Nghề nghiệp và kết quả đăng ký
sns.countplot(
    data=df,
    x="job",
    hue="y",
    ax=axes[0]
)
axes[0].set_title("Job and Subscription")
axes[0].set_xlabel("Nghề nghiệp")
axes[0].set_ylabel("Số lượng khách hàng")
axes[0].tick_params(axis="x", rotation=45)
axes[0].legend(title="Đăng ký")

# Biểu đồ 2: Tình trạng hôn nhân và kết quả đăng ký
sns.countplot(
    data=df,
    x="marital",
    hue="y",
    ax=axes[1]
)
axes[1].set_title("Marital Status and Subscription")
axes[1].set_xlabel("Tình trạng hôn nhân")
axes[1].set_ylabel("Số lượng khách hàng")
axes[1].legend(title="Đăng ký")

plt.tight_layout()
plt.show()
```
