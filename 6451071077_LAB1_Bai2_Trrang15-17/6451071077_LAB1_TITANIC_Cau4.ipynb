```python

print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

# linear algebra
import numpy as np

# data processing
import pandas as pd

# data visualization
import seaborn as sns
%matplotlib inline
from matplotlib import pyplot as plt
from matplotlib import style

# Đọc dữ liệu từ file titanic.csv
with open('titanic.csv', 'r') as f:
    df = pd.read_csv(
        f,
        dtype={
            'age': np.float64,
            'sibsp': np.int64,
            'parch': np.int64
        }
    )
```

```python
print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

FacetGrid = sns.FacetGrid(df, row='embarked', height=4.5, aspect=1.6)
FacetGrid.map(sns.pointplot, 'pclass', 'survived', 'sex', palette=None, order=None, hue_order=None )
FacetGrid.add_legend()

```

```python

print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

plt.figure(figsize=(8, 6))
sns.barplot(x='pclass', y='survived', hue='pclass', data=df, palette=['blue', 'orange', 'green'])
plt.show()
```

```python
print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

grid = sns.FacetGrid(df, col='survived', row='pclass', height=4.2, aspect=1.6)
grid.map(plt.hist, 'age', alpha =.5, bins=20)
grid.add_legend();
```
