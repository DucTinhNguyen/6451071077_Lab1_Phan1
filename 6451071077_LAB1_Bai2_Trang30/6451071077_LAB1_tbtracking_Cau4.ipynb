```python

print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

# Linear algebra
import numpy as np

# Data processing
import pandas as pd

# Data visualization
import seaborn as sns
%matplotlib inline
from matplotlib import pyplot as plt
from matplotlib import style

# Đọc dữ liệu đã xử lý từ file tb_tracking_processed.csv
df = pd.read_csv(
    'tb_tracking_processed.csv',
    parse_dates=['time']
)
```

```python
print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

FacetGrid = sns.FacetGrid(df, row='device_id', height=4.5, aspect=1.6)
FacetGrid.map(sns.pointplot, 'minutes', 'co_level', 'humidity',
              palette=None, order=None, hue_order=None)
FacetGrid.add_legend()
```

```python
print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

sns.barplot(x='device_id', y='co_level', data=df)
```

```python
print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

grid = sns.FacetGrid(df, col='device_id', height=4.2, aspect=1.6)
grid.map(plt.hist, 'temperature', alpha=.5, bins=20)
grid.add_legend()
```
