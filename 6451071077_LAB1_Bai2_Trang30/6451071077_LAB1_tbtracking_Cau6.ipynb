```python
print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

for dataset in [df]:
    mean = df['co_level'].mean()
    std = df['co_level'].std()
    is_null = dataset['co_level'].isnull().sum()
    # compute random numbers between the mean, std and is_null
    rand_co = np.random.randint(mean - std, mean + std, size=is_null)
    # fill NaN values in co_level column with random values generated
    co_slice = dataset['co_level'].copy()
    co_slice[np.isnan(co_slice)] = rand_co
    dataset['co_level'] = co_slice

df['co_level'].isnull().sum()
```
